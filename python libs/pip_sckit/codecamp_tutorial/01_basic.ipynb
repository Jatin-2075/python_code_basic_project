```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
coln = ["fLenght", "fWidth", "fSize", "fConel", "fAsym", "fM3Long", "fM3Trans", "fAlpha", "fDist", "class" ]
df = pd.read_csv("magic04.data", names=coln)
df.head()
```

```python
df["class"] = (df["class"] == 'g').astype(int)
```

```python
df.head()
```

```python
for label in coln[:-1]:
    plt.hist(df[df["class"] == 1][label], color='blue', alpha=0.7, density=True, label='gamma')
    plt.hist(df[df["class"] == 0][label], color='red', alpha=0.7, density=True, label='beta')
    plt.title(label)
    plt.xlabel(label)
    plt.ylabel("Probability")
    plt.legend()
    plt.show()
```
