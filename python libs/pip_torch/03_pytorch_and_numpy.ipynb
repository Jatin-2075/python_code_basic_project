To make use of a GPU (if available), we first need to check for CUDA availability and define our `device`.

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")
```

Now, let's create a tensor and move it to the `device` (GPU if available, otherwise CPU).

```python
# Create a tensor on the CPU
some_tensor = torch.tensor([1, 2, 3])
print(f"Tensor on CPU: {some_tensor}, device: {some_tensor.device}")

# Move the tensor to the target device
some_tensor_on_device = some_tensor.to(device)
print(f"Tensor on {device}: {some_tensor_on_device}, device: {some_tensor_on_device.device}")
```

To move a tensor back to the CPU, you can use the `.cpu()` method. This is often necessary when you want to use NumPy with your tensor, as NumPy can only handle CPU tensors.

```python
# Move the tensor back to the CPU
some_tensor_on_cpu = some_tensor_on_device.cpu()
print(f"Tensor back on CPU: {some_tensor_on_cpu}, device: {some_tensor_on_cpu.device}")

# You can now convert it to NumPy array
numpy_array_from_tensor = some_tensor_on_cpu.numpy()
print(f"NumPy array from tensor: {numpy_array_from_tensor}, type: {type(numpy_array_from_tensor)}")
```

When performing operations with tensors, PyTorch will automatically use the device (GPU or CPU) where the tensors are located. Let's see an example of a tensor operation on the chosen `device` (which is `cpu` in this current runtime) and then explicitly on the `cpu`.

```python
# Create two tensors directly on the defined 'device'
tensor_a = torch.tensor([10., 20., 30.]).to(device)
tensor_b = torch.tensor([1., 2., 3.]).to(device)

# Perform an operation (addition) on the device
result_on_device = tensor_a + tensor_b

print(f"Tensor A on {tensor_a.device}: {tensor_a}")
print(f"Tensor B on {tensor_b.device}: {tensor_b}")
print(f"Result of A + B on {result_on_device.device}: {result_on_device}")
```

If you want to ensure an operation is performed on the CPU, even if a GPU is available, you can explicitly move the tensors to the CPU before the operation.

```python
# Create two tensors on the CPU explicitly
cpu_tensor_a = torch.tensor([5., 10., 15.]).cpu()
cpu_tensor_b = torch.tensor([0.5, 1.0, 1.5]).cpu()

# Perform an operation (multiplication) on the CPU
cpu_result = cpu_tensor_a * cpu_tensor_b

print(f"CPU Tensor A on {cpu_tensor_a.device}: {cpu_tensor_a}")
print(f"CPU Tensor B on {cpu_tensor_b.device}: {cpu_tensor_b}")
print(f"Result of A * B on {cpu_result.device}: {cpu_result}")
```

### Important Note on GPU Usage

As observed earlier, your current Colab runtime is using the CPU (`Using device: cpu`). To utilize a GPU for these operations, you need to change your Colab runtime type.

**To enable GPU in Colab:**
1. Go to `Runtime` in the top menu.
2. Select `Change runtime type`.
3. In the `Hardware accelerator` dropdown, choose `GPU`.
4. Click `Save`.

After changing the runtime, you will need to re-run the notebook cells, starting from where `torch` is imported, to ensure the `device` variable is correctly set to `cuda`.
